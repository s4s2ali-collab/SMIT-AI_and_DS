# **Q8:** Ek list `numbers = [12, 45, 7, 23, 56, 90, 11]` mein se sabse bada (Maximum) number find karein loop ke zariye.

````python
# type your code here

numbers = [12, 45, 7, 23, 56, 90, 11]
max_num = numbers[0]
index = 0

while index < len(numbers):
    if numbers[index] > max_num:
        max_num = numbers[index]
    index+=1
    
print(max_num)
````

# **Q12:** User se ek number lein aur check karein ke wo **Prime Number** hai ya nahi (using loops).

````python
# type your code here

pm_check = int(input("Enter any number of your choice: "))

is_pm = True

if pm_check <= 1:
    is_pm = False
else:
    lm = int(pm_check**0.5) + 1
    
for p in range(2, lm):
    if pm_check % 1 == 0:
        is_pm = False
        break

if is_pm:
    print(f"{pm_check} is a Prime Number")
else:
    print(f"{pm_check} is not a Prime Number")
````

# **Q13:** Nested loop ka use karke niche diya gaya pattern print karein:
```
*
**
***
****
*****
```

````python
# type your code here

rows = 5

for s in range(1, rows+s):
    space_s = " " *(rows - s)
    star_s = "*" * (2*s-1)
    print(space_s + star_s)

````

# **Q16:** User se ek number input lein aur us ka **Factorial** find karein (e.g., 5! = 5*4*3*2*1).

````python
# type your code here

f_num = int(input("Enter any number of your choice to find Factorial: "))
factorial_f = 1
f = f_num

while f > 0:
    factorial_f *= f
    f -= 1

print(f"The factorial of {f_num} is: {factorial_f}")

````

# **Q18:** Ek manual counting karein ke ek string (e.g., 'Learning Python is fun') mein kitne **Vowels** (a, e, i, o, u) hain.

````python
# type your code here

text_a = "Learning Python is fun"
vowels = "aeiou"
count_a = 0

for char in text_a:
    if char.lower() in vowels:
        count_a += 1
print(f"Number of vowels are {count_a}.")
````

# **Q20:** Ek 'Number Guessing Game' banayein jahan program ek number (1-20) fix kare aur user ko 3 chances de guess karne ke liye. `while loop` aur `break` ka use karein.

````python
# type your code here

import random

lucky_num = random.randint(1, 20)
attempts_a = 0
max_try = 3

print("I've selected a number between 1 and 20. You have 3 attempts to guess it!")

won = False

while attempts_a < max_try:
    try:
        guess = int(input(f"Tries {attempts_a + 1}: Enter your guess: "))
        attempts_a += 1
        
        if guess == lucky_num:
            print(f"Congratulations! You guessed the correct number in {attempts_a} tries.")
            won = True  # Set flag to True
            break
        else:
            print("Better luck next time.")
            
    except ValueError:
        print("Invalid input! Please enter a valid whole number.")

if not won:
    print(f"Game over! The lucky number was {lucky_num}.")

````
