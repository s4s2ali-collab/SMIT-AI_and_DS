**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2

class BankAccount:
    def __init__(self, AccountNumber, ID, balance=0):
        self.AccountNumber = AccountNumber
        self.ID = ID
        self.__balance = int(balance)

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: PKR {amount} ; New balance: PKR {self.__balance}")
        else:
            print(f"Error! Deposit amount should be a positive number.")

    def withdraw(self,amount):
        if amount <= 0:
            print("Error! Withdrawal amount must be positive number")
            return
        if amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: PKR {amount} ; Remaining balance: PKR {self.__balance}")
        else:
            print(f"Transaction Declined: Insufficient balance.")
            
```

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
# Write your code here for Q7

class Animal:
    def __init__(self, kingdom):
        self.kingdom = kingdom

    def breathe(self):
        return "Breathing oxygen..."

class Mammal(Animal):
    def __init__(self, kingdom, fur_color):
        super().__init__(kingdom)
        self.fur_color = fur_color

    def feed_milk(self):
        return "Feeding milk to young..."

class Dog(Mammal):
    def __init__(self, kingdom, fur_color, breed):
        super().__init__(kingdom, fur_color)
        self.breed = breed

    def bark(self):
        return "Woof! Woof!"
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10

# Hierarchical Inheritance occurs when multiple child classes inherit from a single parent class. It creates a tree-like structure where one base class shares its common properties and methods with two or more specialized sub-classes

class Shape:
    def __init__(self, color: str):
        self.color = color

    def display_color(self):
        print(f"This shape is colored {self.color}.")


class Circle(Shape):
    def __init__(self, color: str, radius: float):
        super().__init__(color)
        self.radius = radius

    def area(self) -> float:
        return 3.14159 * (self.radius ** 2)


class Square(Shape):
    def __init__(self, color: str, side: float):
        super().__init__(color)
        self.side = side

    def area(self) -> float:
        return self.side * self.side


if __name__ == "__main__":
    print("--- Testing Circle (Child 1) ---")
    my_circle = Circle(color="Red", radius=5.0)
    my_circle.display_color()
    print(f"Circle Area: {my_circle.area():.2f}")

    print("\n--- Testing Square (Child 2) ---")
    my_square = Square(color="Blue", side=4.0)
    my_square.display_color()
    print(f"Square Area: {my_square.area():.2f}")


```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Write your code here for Q11

# Method overriding is an Object-Oriented Programming (OOP) feature that allows a child class to provide a specific implementation of a method that is already defined in its parent class
# When a method in a child class has the same name, same parameters, and same return type as a method in its parent class, the child class method overrides (replaces) the parent method.

class Bird:
    def fly(self):
        print("Most birds can fly")


class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly")


if __name__ == "__main__":
    generic_bird = Bird()
    generic_bird.fly()

    ostrich = Ostrich()
    ostrich.fly()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13

# Operator Overloading allows you to change how built-in operators (like +, -, *, or /) behave when used with custom user-defined objects.In Python, operators work by calling special "magic methods" or "dunder methods" (double underscore methods) behind the scenes. When you write object1 + object2, Python automatically translates it into object1.__add__(object2)

class ComputerInventory:
    def __init__(self, count: int):
        self.count = count

    def __add__(self, other):
        if isinstance(other, ComputerInventory):
            return ComputerInventory(self.count + other.count)
        raise TypeError("You can only add a ComputerInventory to another ComputerInventory")

    def __str__(self):
        return f"Total Computers: {self.count}"


if __name__ == "__main__":
    laptop_stock = ComputerInventory(15)
    desktop_stock = ComputerInventory(25)

    total_stock = laptop_stock + desktop_stock

    print(total_stock)


```

```python
# Write your code here for Q14

# The built-in len() function shows polymorphism because it accepts different object types (Strings, Lists, Dictionaries) and behaves differently for each, yet provides a consistent interface to the user. Under the hood, Python uses "duck typing"—whenever you call len(obj), Python automatically calls the object's internal __len__() dunder method, allowing each class to define how its own size is calculated.

my_string = "Python OOP"
my_list = [10, 20, 30, 40]
my_dict = {"brand": "Dell", "ram": 16, "storage": 512}

for data_structure in (my_string, my_list, my_dict):
    print(f"Type: {type(data_structure).__name__:<10} | Length: {len(data_structure)}")


```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Write your code here for Q15


# No, Python does not natively support method overloading (having the same method name with different numbers or types of arguments in the same class). If you define multiple methods with the exact same name, Python will simply overwrite the old ones and only remember the very last definition
# To handle same-name methods with multiple arguments, we use a single method with default arguments, variable-length arguments (*args or **kwargs), or the functools.singledispatchmethod decorator

from functools import singledispatchmethod

class Computer:
    def boot(self, OS: str = "Windows"):
        print(f"Booting computer into {OS}...")

    def calculate(self, *args: int) -> int:
        if len(args) == 0:
            return 0
        elif len(args) == 1:
            return args[0] ** 2
        else:
            return sum(args)

    @singledispatchmethod
    def connect(self, device):
        raise TypeError("Unsupported device type")

    @connect.register(str)
    def _(self, device: str):
        print(f"Connecting to Wi-Fi network: {device}")

    @connect.register(int)
    def _(self, device: int):
        print(f"Connecting to Bluetooth device ID: {device}")


if __name__ == "__main__":
    pc = Computer()

    pc.boot()
    pc.boot("Linux")

    print(pc.calculate())
    print(pc.calculate(5))
    print(pc.calculate(10, 20, 30))

    pc.connect("Home_WiFi")
    pc.connect(42)


```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


class Laptop(Computer):
    def process(self):
        print("Laptop is processing data using a mobile, power-efficient CPU.")


class Desktop(Computer):
    def process(self):
        print("Desktop is processing data using a high-performance desktop CPU.")


if __name__ == "__main__":
    my_laptop = Laptop()
    my_laptop.process()

    my_desktop = Desktop()
    my_desktop.process()

```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18

from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        print("Engine started. Vroom vroom!")

    @abstractmethod
    def drive(self):
        pass


class ElectricCar(Car):
    def drive(self):
        print("Driving smoothly and silently on electric power.")


if __name__ == "__main__":
    my_car = ElectricCar()
    my_car.start_engine()
    my_car.drive()
```

**Q19:** Abstraction aur Encapsulation ke darmiyan kya wazeh farq (main difference) hai? Dono ke concepts ko short statements mei define karein.

####### Write your code here for Q19

# The core difference:
  - Abstraction hides complexity (shows what an object does, not how).
  - Encapsulation hides data (restricts who can see or change internal variables)

# Abstraction
  - Showing only the essential features of an object to the user while hiding the messy background details
  - Example: Using @abstractmethod. The user calls .process_payment() without needing to understand the complex bank gateway APIs running under the hood.

# Encapsulation
  - Bundling data (variables) and methods together into a single unit (a class) and restricting direct access to prevent accidental tampering
  - Example: Using double underscores __ to make a variable private. Making self.__cvv private so external code cannot view or modify the credit card's security code directly

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20

from abc import ABC, abstractmethod

class Payment(ABC):
    def __init__(self, amount: float):
        self.amount = amount

    @abstractmethod
    def process_payment(self) -> bool:
        pass

    def generate_receipt(self, transaction_id: str, success: bool):
        if success:
            print(f"Receipt: Successfully processed ${self.amount:.2f}. Transaction ID: {transaction_id}")
        else:
            print(f"Receipt: Payment of ${self.amount:.2f} failed.")


class CreditCardPayment(Payment):
    def __init__(self, amount: float, card_number: str, cvv: str):
        super().__init__(amount)
        self.card_number = card_number
        self.cvv = cvv

    def process_payment(self) -> bool:
        print(f"Connecting to Card Gateway... Verifying card ending in {self.card_number[-4:]}...")
        if len(self.cvv) == 3:
            print("Authorization successful.")
            return True
        print("Authorization failed: Invalid CVV.")
        return False


class UPIPayment(Payment):
    def __init__(self, amount: float, upi_id: str):
        super().__init__(amount)
        self.upi_id = upi_id

    def process_payment(self) -> bool:
        print(f"Sending collect request of ${self.amount:.2f} to UPI ID: {self.upi_id}...")
        print("Awaiting user PIN confirmation in their banking app...")
        print("PIN verified successfully.")
        return True


if __name__ == "__main__":
    print("--- Scenario 1: User chooses Credit Card ---")
    card_tx = CreditCardPayment(amount=125.50, card_number="4111222233334444", cvv="999")
    is_card_success = card_tx.process_payment()
    card_tx.generate_receipt(transaction_id="TXN_CARD_78910", success=is_card_success)

    print("\n--- Scenario 2: User chooses UPI ---")
    upi_tx = UPIPayment(amount=45.00, upi_id="user@bank")
    is_upi_success = upi_tx.process_payment()
    upi_tx.generate_receipt(transaction_id="TXN_UPI_11121", success=is_upi_success)

```
