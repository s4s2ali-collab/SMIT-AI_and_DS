### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
df_clean = movie_data.dropna(subset=['Certificate'])
cert_gross = df_clean.groupby('Certificate')['Gross'].mean().reset_index()
cert_gross = cert_gross.sort_values(by='Gross', ascending=False)

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))
ax = sns.barplot(data=cert_gross, x="Certificate", y="Gross", color="royalblue")

for p in ax.patches:
    val = p.get_height()
    if val > 0:
        label = f"${val/1e6:.1f}M" if val >= 1e6 else f"${val:,.0f}"
        ax.annotate(label, (p.get_x() + p.get_width() / 2., val),
                    ha='center', va='baseline', fontsize=10, 
                    fontweight='bold', xytext=(0, 5), textcoords='offset points')

plt.title("Average Gross Earnings by Movie Certificate", fontsize=12, fontweight="bold")
plt.ylim(0, cert_gross['Gross'].max() * 1.15)
plt.tight_layout()
plt.show()


```

### **Question 27**
(Visualization) Har saal (Released_Year) kitni movies release hueen, iska ek Line Plot banayein.

```python
clean_years = pd.to_numeric(movie_data['Released_Year'], errors='coerce').dropna().astype(int)
yearly_counts = clean_years.value_counts().sort_index()

sns.set_theme(style="whitegrid")
plt.figure(figsize=(16, 6))

ax = sns.lineplot(
    x=yearly_counts.index, 
    y=yearly_counts.values, 
    color="royalblue", 
    marker="o", 
    linewidth=2.5
)

plt.xticks(ticks=yearly_counts.index, labels=yearly_counts.index, rotation=90, fontsize=9)

if len(yearly_counts) <= 25:
    for x, y in zip(yearly_counts.index, yearly_counts.values):
        ax.text(x, y, f"{int(y)}", ha='center', va='bottom', 
                fontsize=9, fontweight='bold', color='black', xytext=(0, 5), textcoords='offset points')

plt.title("Total Number of Movies Released per Year", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Released Year", fontsize=12)
plt.ylabel("Number of Movies", fontsize=12)

plt.ylim(0, yearly_counts.max() * 1.1)

plt.tight_layout()
plt.show()
```

### **Question 28**
(Visualization) 'No_of_Votes' ke mutabiq top 5 highest voted movies ka ek Horizontal Bar Chart banayein.

```python
top_voted = movie_data.nlargest(5, 'No_of_Votes')

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

ax = sns.barplot(
    data=top_voted,
    x="No_of_Votes",
    y="Series_Title",
    color="royalblue"
)

for p in ax.patches:
    width = p.get_width()
    if width > 0:
        if width >= 1_000_000:
            label = f"{width / 1_000_000:.2f}M"
        elif width >= 1_000:
            label = f"{width / 1_000:.1f}K"
        else:
            label = f"{width:,.0f}"
            
        ax.annotate(
            label, 
            (width, p.get_y() + p.get_height() / 2.),
            ha='left',
            va='center',
            fontsize=10, 
            fontweight='bold', 
            color='black', 
            xytext=(7, 0),
            textcoords='offset points'
        )

plt.title("Top 5 Highest Voted Movies", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Number of Votes", fontsize=12)
plt.ylabel("Movie Title", fontsize=12)

ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f"{x*1e-6:.1f}M" if x >= 1e6 else f"{x:,.0f}"))

plt.xlim(0, top_voted['No_of_Votes'].max() * 1.15)

plt.tight_layout()
plt.show()

```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
movie_data['IMDB_Rating_Scaled'] = movie_data['IMDB_Rating'] * 10

top_meta = movie_data.nlargest(5, 'Meta_score')
top_imdb = movie_data.nlargest(5, 'IMDB_Rating_Scaled')

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.barplot(
    data=top_meta, 
    x="Meta_score", 
    y="Series_Title", 
    ax=axes[0], 
    color="royalblue"
)
axes[0].set_title("Top 5 Movies by Metascore (Out of 100)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Metascore", fontsize=11)
axes[0].set_ylabel("Movie Title", fontsize=11)
axes[0].set_xlim(0, 115)

for p in axes[0].patches:
    width = p.get_width()
    if width > 0:
        axes[0].annotate(f"{width:.0f}", (width, p.get_y() + p.get_height() / 2.),
                         ha='left', va='center', fontsize=10, fontweight='bold',
                         xytext=(5, 0), textcoords='offset points')

sns.barplot(
    data=top_imdb, 
    x="IMDB_Rating_Scaled", 
    y="Series_Title", 
    ax=axes[1], 
    color="crimson"
)
axes[1].set_title("Top 5 Movies by IMDb Rating (Scaled to 100)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("IMDb Rating (Scaled)", fontsize=11)
axes[1].set_ylabel("")
axes[1].set_xlim(0, 115)

for p in axes[1].patches:
    width = p.get_width()
    if width > 0:
        axes[1].annotate(f"{width:.1f}", (width, p.get_y() + p.get_height() / 2.),
                         ha='left', va='center', fontsize=10, fontweight='bold',
                         xytext=(5, 0), textcoords='offset points')

plt.tight_layout()
plt.show()


```
